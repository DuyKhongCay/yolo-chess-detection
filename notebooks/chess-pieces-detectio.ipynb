```python
from ultralytics import YOLO

# Load pretrained YOLOv11m checkpoint
model = YOLO("/kaggle/input/models/duykhongcay/chesspieces-yolo11/pytorch/default/1/best.pt")

# Train fine-tuning model directly inside notebook cell
results = model.train(
    data=str(LOCAL_DIR / "data.yaml"),
    project=RUNS_DIR,
    name="chess_detection_yolo11n",
    epochs=30,
    batch=16,
    imgsz=640,
    device=[0,1],
    optimizer="AdamW",
    lr0= 0.001,
    mosaic=0.5,
    erasing=0.0,
    scale=0.2,
    exist_ok=False,
    cache=False,
    resume=False,
    patience=10
)
```
